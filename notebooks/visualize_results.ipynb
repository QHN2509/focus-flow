# Visualize training artifacts

This notebook loads the artifacts produced by `train_classifier.py` in `artifacts/` and visualizes the confusion matrix, class distribution, and cross-validation summary.

```python
import json
import os
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
from PIL import Image

ROOT = Path('..').resolve() if Path('.').name == 'notebooks' else Path('.')
ARTIFACTS = ROOT / 'artifacts'

# Find the latest metadata file by timestamp in filename
meta_files = sorted(ARTIFACTS.glob('metadata_*.json'))
if not meta_files:
    print('No metadata files found in', ARTIFACTS)
else:
    meta_path = meta_files[-1]
    with open(meta_path, 'r', encoding='utf-8') as f:
        metadata = json.load(f)
    print('Loaded metadata from', meta_path)
    display(metadata)
```

```python
# Show confusion matrix image if present
cm_path = ARTIFACTS / metadata.get('confusion_matrix', 'confusion_matrix.png')
if cm_path.exists():
    display(Image.open(cm_path))
else:
    print('Confusion matrix not found at', cm_path)
```

```python
# Plot class distribution
if 'class_distribution' in metadata:
    cd = metadata['class_distribution']
    labels = list(cd.keys())
    counts = list(cd.values())
    fig, ax = plt.subplots(figsize=(6,4))
    ax.bar(labels, counts, color='tab:blue')
    ax.set_title('Class distribution (from metadata)')
    ax.set_ylabel('Count')
    plt.show()
else:
    print('No class distribution in metadata')
```

```python
# Show CV summary if present
cv_path = ARTIFACTS / metadata.get('cv_summary', 'cv_summary.json')
if cv_path.exists():
    with open(cv_path, 'r', encoding='utf-8') as f:
        cv = json.load(f)
    print('Cross-validation summary:')
    display(cv)
else:
    print('CV summary not found at', cv_path)
```
